```python
import pickle
from pathlib import Path

from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS
```

```python
PROJECT_ROOT = Path.cwd().parent

processed_dir = PROJECT_ROOT / "data" / "processed"
vector_store_dir = PROJECT_ROOT / "vector_store"

vector_store_dir.mkdir(parents=True, exist_ok=True)
```

```python
with open(processed_dir / "cv_documents.pkl", "rb") as f:
    cv_documents = pickle.load(f)

with open(processed_dir / "jd_documents.pkl", "rb") as f:
    jd_documents = pickle.load(f)
```

```python
print("CV Documents:", len(cv_documents))
print("JD Documents:", len(jd_documents))
```

```python
embedding_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2",
    model_kwargs={"device": "cpu"},
    encode_kwargs={"normalize_embeddings": True}
)
```

```python
cv_vectorstore = FAISS.from_documents(
    cv_documents,
    embedding_model
)
```

```python
jd_vectorstore = FAISS.from_documents(
    jd_documents,
    embedding_model
)
```

```python
cv_vectorstore.save_local(vector_store_dir / "cv_index")

jd_vectorstore.save_local(vector_store_dir / "jd_index")
```

```python
cv_retriever = cv_vectorstore.as_retriever(
    search_kwargs={"k": 3}
)

jd_retriever = jd_vectorstore.as_retriever(
    search_kwargs={"k": 3}
)
```

```python
query = "What programming languages does the candidate know?"

docs = cv_retriever.invoke(query)

for i, doc in enumerate(docs, start=1):
    print(f"\nDocument {i}")
    print("-" * 50)
    print(doc.page_content)
```

```python
query = "What skills are required for this job?"

docs = jd_retriever.invoke(query)

for i, doc in enumerate(docs, start=1):
    print(f"\nDocument {i}")
    print("-" * 50)
    print(doc.page_content)
```
